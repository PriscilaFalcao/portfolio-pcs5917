```python
import kagglehub
from pathlib import Path
# soluçao para baixaar apenas os 100 primeiros
arquivos = []

for indice in range(100):
    nome = f"T_{indice:010d}.flac"
    caminho = kagglehub.dataset_download(
        "ulianazb/asvspoof-2024",
        path=f"flac_T/flac_T/{nome}"
    )
    arquivo = Path(caminho)
    if not arquivo.is_file():
        raise FileNotFoundError(f"Arquivo nao encontrado: {arquivo}")
    arquivos.append(arquivo)
    print(f"{indice + 1}/100: {arquivo}")

print(f"Arquivos disponiveis: {len(arquivos)}")
print("Pasta local:", arquivos[0].parent)
```

```python
from pathlib import Path
import pandas as pd
import soundfile as sf
from tqdm.auto import tqdm
from pyAudioAnalysis import ShortTermFeatures

if not arquivos:
    raise ValueError("Execute primeiro a celula que baixa os 100 arquivos")

raiz = Path(arquivos[0]).parent

registros = []
erros = []

for arquivo in tqdm(arquivos):
    try:
        sinal, fs = sf.read(arquivo, always_2d=True)
        sinal = sinal.mean(axis=1)
        janela = int(0.050 * fs)
        passo = int(0.025 * fs)

        if len(sinal) < janela:
            raise ValueError("Audio menor que a janela de 50 ms")

        features, nomes = ShortTermFeatures.feature_extraction(
            sinal, fs, janela, passo, deltas=False
        )

        registro = {
            "arquivo": str(arquivo.relative_to(raiz)),
            "audio_id": arquivo.stem,
            "duracao_s": len(sinal) / fs,
            "sample_rate": fs,
        }
        for nome, media, desvio in zip(
            nomes, features.mean(axis=1), features.std(axis=1)
        ):
            registro[f"{nome}_mean"] = media
            registro[f"{nome}_std"] = desvio

        registros.append(registro)
    except Exception as erro:
        erros.append({"arquivo": str(arquivo), "erro": str(erro)})

df = pd.DataFrame(registros)
display(df.head())
display(pd.DataFrame(erros))
df.to_csv("audio_features.csv", index=False)
print(f"Tabela gerada: {df.shape[0]} audios, {df.shape[1]} colunas; erros: {len(erros)}")
```
